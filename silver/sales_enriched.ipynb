## Requirement 1: the merge, stated explicitly

The channel features are physically resolved in `gold.dim_channel`, which is
where they belong: they are attributes of the channel, so they are joined
once against 31 rows instead of being carried on every fact row.

The case wording, however, asks for the channel features to be merged with
the transactional sales. This view states that merge directly, so the
requirement is visible without having to reconstruct it from the star schema.

It is a view rather than a table on purpose. It duplicates no data, it cannot
drift from its sources, and it costs nothing to maintain.

```python
CATALOG_NAME = 'beverage_sales'
SCHEMA_NAME = 'silver'
VIEW_NAME = 'sales_enriched'
```

```python
spark.sql(f"""
CREATE OR REPLACE VIEW {CATALOG_NAME}.{SCHEMA_NAME}.{VIEW_NAME} AS
SELECT
    s.*,
    c.trade_group,
    c.trade_type
FROM {CATALOG_NAME}.{SCHEMA_NAME}.sales s
LEFT JOIN {CATALOG_NAME}.{SCHEMA_NAME}.channel_group c
  ON s.trade_channel = c.trade_channel
""")
```

```python
display(spark.table(f'{CATALOG_NAME}.{SCHEMA_NAME}.{VIEW_NAME}'))
```
